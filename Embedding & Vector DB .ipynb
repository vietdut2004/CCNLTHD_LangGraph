```python
import os
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

# Dữ liệu
sentences = [
    "Tôi cảm thấy lo lắng",
    "Tôi rất vui hôm nay",
    "Công việc thật áp lực",
    "Tôi thấy thật bình yên",
    "Tôi hạnh phúc"
]

#  Tạo embedding model
# model_name = "VoVanPhuc/sup-SimCSE-VietNamese-phobert-base"
model_name = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"

print(f"Đang tải model embedding local: '{model_name}'...")
# THÊM: normalize_embeddings=True để tự động normalize
embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs={'device': 'cpu'},
    encode_kwargs={'normalize_embeddings': True}  #  QUAN TRỌNG!
)
print("Đã tải model xong!")

#  Tạo FAISS vector store
print("Đang tạo vector store...")
db = FAISS.from_texts(
    sentences,
    embedding=embeddings
)
print("Đã tạo vector store xong!")

#  Thực hiện semantic search
query = "Những cảm xúc tích cực"
docs_and_scores = db.similarity_search_with_score(query, k=3)

#  In kết quả
print(f"\n--- Top 3 kết quả cho truy vấn: '{query}' ---\n")
for i, (doc, score) in enumerate(docs_and_scores, 1):
    print(f"{i}. {doc.page_content} — (Score: {score:.4f})")

print("\n-------------------------------------------------")
print("\n PHÂN TÍCH:")
print(" Score càng NHỎ (gần 0) = Câu càng GIỐNG với truy vấn")
print(" Score càng LỚN (gần 2) = Câu càng XA với truy vấn")
```
